JOINS 

```python
import pandas as pd
```

```python
#sample dataframes
employees = pd.DataFrame({
    'name': ['Sindhu', 'Rahul', 'Priya'],
    'department': ['IT', 'HR', 'Finance']
}, index = [101, 102, 103])
```

```python
print(employees)
```

```python
salary = pd.DataFrame({
    'salary': [50000, 60000, 70000]
}, index = [101, 102, 104])
```

```python
print(salary)
```

Left Join

```python
result = employees.join(salary, how = 'left')
print(result)
```

Right Join

```python
result = employees.join(salary, how = 'right')
result
```

Inner Join

```python
result = employees.join(salary, how = 'inner')
result
```

Outer Join

```python
result = employees.join(salary, how = 'outer')
result
```

Join using a Column and an Index

```python
customers = pd.DataFrame({
    'customer_id': [101, 102, 103],
    'customer_name': ['sindhu', 'rahul', 'priya']
})

orders = pd.DataFrame({
    'amount': [500, 1000, 1500],
}, index = [101, 102, 103])
```

```python
result = customers.join(
    orders,
    on = 'customer_id',
    how = 'left'
)

result
```

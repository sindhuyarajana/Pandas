```python
#filtering using condition operators
pd.set_option('display.max.rows', 240)
```

```python
df[df['Units Sold'] <= 5]
```

```python
#filtering text/string cols
df[df['Product Category'] == 'Electronics']
```

```python
#filtering using isin function
df[df['Product Category'].isin(['Books', 'Clothing'])]
```

```python
# sorting in descending order
df.sort_values("Units Sold", ascending = False).head(5)
```

```python
#sorting by multiple columns
df.sort_values(by = ['Units Sold', 'Unit Price'], ascending = [False, False]). head(5)
```

```python
#sorting by indexing
df.sort_index(ascending = False)
```

```python
#combining filtering and ordering
df[df['Product Category'] == 'Beauty Products'].sort_values('Units Sold', ascending = False)
```

```python
df[(df['Product Category'] == 'Home Appliances') & (df['Units Sold'] > 1)].sort_values("Total Revenue", ascending = True)
```

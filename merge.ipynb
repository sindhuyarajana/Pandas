```python
import pandas as pd
```

CREATING SAMPLE DATAFRAMES

```python
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'customer_name': ['Sindhu', 'Priti', 'Trishita', 'Nilamesh'],
    'city': ['Mumbai', 'Delhi', 'Chennai', 'Kolkata']
})
```

```python
orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104],
    'customer_id': [1, 2, 2, 5],
    'amount': [500, 1000, 1500, 2000]
})
```

Inner Merge

```python
result1 = pd.merge(
    customers, 
    orders,
    on = 'customer_id',
    how = 'inner'
)

print(result1)
```

```python
result1.set_index('customer_id')
```

Left Merge

```python
result2 = pd.merge(
    customers,
    orders,
    on = 'customer_id',
    how = 'left'
)

print(result2)
```

Rightr Merge

```python
result3 = pd.merge(
    customers,
    orders,
    on = 'customer_id',
    how = 'right'
)

print(result3)
```

Outer Merge

```python
result4 = pd.merge(
    customers,
    orders,
    on = 'customer_id',
    how = 'outer'
)

print(result4)
```

Cross Merge

```python
result5 = pd.merge(
    customers,
    orders,
    how = 'cross'
)
print(result5)
```

Merge using different column names

```python
customers = pd.DataFrame({
    'customer_id': [1, 2, 3],
    'customer_name': ['Sindhu', 'Priti', 'Trishita']
}) 

orders = pd.DataFrame({
    'client_id': [1, 2, 4],
    'order_amount': [500, 1000, 1500]
})
```

```python
result = pd.merge(
    customers,
    orders,
    left_on = 'customer_id',
    right_on = 'client_id',
    how = 'inner'
)

print(result)
```

Merge using Multiple Columns

```python
df1 = pd.DataFrame({
    'customer_id': [1, 1, 2],
    'year': [2025, 2026, 2025],
    'sales': [1000, 2000, 1500]
})

df2 = pd.DataFrame({
    'customer_id': [1, 1, 2],
    'year': [2025, 2026, 2025],
    'target': [900, 1800, 1400]
})
```

```python
result = pd.merge(
    df1,
    df2,
    on = ['customer_id', 'year'],
    how = 'inner'
)

print(result)
```

Handling duplicate column names using suffixes

```python
df1 = pd.DataFrame({
    'customer_id': [1, 2],
    'name': ['Sindhu', 'Priti'],
    'city': ['Mumbai', 'Delhi']
})

df2 = pd.DataFrame({
    'customer_id': [1, 2],
    'name': ['S.Sindhu', 'P.Priti'],
    'amount': [500, 1000]
})
```

```python
result = pd.merge(
    df1,
    df2,
    on = 'customer_id'
)
print(result)
```

```python
result = pd.merge(
    df1,
    df2,
    on = 'customer_id',
    suffixes = ('_customer', '_order')
)
print(result)
result.set_index('customer_id')
```

Using indicator = True

```python
result = pd.merge(
    customers,
    orders,
    left_on = 'customer_id',
    right_on = 'client_id',
    how = 'outer',
    indicator = True
)
print(result)
```

Using validate parameter

```python
result = pd.merge(
    customers,
    orders,
    left_on = 'customer_id',
    right_on = 'client_id',
    how = 'left',
    validate = 'one_to_many'
)

print(result)
```
